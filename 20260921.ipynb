```python
a = (10,'t',[20,30,40])  # 다양한 자료형 가능
b = (5,)                 # 값이 1개인 튜플 생성
c = tuple(range(3))      # range()로 생성
red = (255,0,0)          # 불변 값 저장에 유용
print(a, type(a))
print(c,len(c))
print(a+b)               # 덧셈 연산자
print(red * 2)           # 곱셈 연산자
print(a[1])              # 인덱싱
print(a[2][1:])          # 슬라이싱
print(a.index('t'))      # 메소드
```
