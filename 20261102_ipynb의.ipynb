```python
for i in range(1,4):         #3줄을 출력하기 위해 횟수 반복
  for j in range(1, i + 1):   #각 줄(라인)에 출력력해야 할 '*'
    print('*', end = '')        #'*' 출력
  print()                      #줄 바꾸기
```
