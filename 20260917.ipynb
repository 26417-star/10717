```python
a =[]  # 공백 리스트
b =[10,20,30]
c =[1,2,'hello', 'hi']
d =[10,2.5,['hello' 'hi']]
print(a, type(a))
print(b, type(b))
print(c, type(c))
print(d, type(d))
```

```python
a =[10,20,30, ['a','b','c']]
b =[40,50,60, ['a','b','c'], 1.5,-5]
print(a[0])
print(a[-1])
print(a[3][0])   #중첩 리스트의 인덱싱
print(b[3][:2])  #중첩 리스트의 슬라이싱

```

```python
a =[4,2,3]
b =[4,5]
c =[1,2,[3,4]]
print(a + b)
print(a * 3)
print(a < b)   # 2 < 5 (true)
print(len(a))
print(len(c))
print(len(c[2]))  # len([3,4])
```

```python
a = [1,2,3,10,20]
print('원본:',a)
a[2] =4   # 1. 인덱싱 수정
print('결과1:',a)
a[3:] =[5,6] # 2. 슬라이싱 수정
print('결과')
```
