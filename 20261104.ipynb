```python
num = int(input('정수입력: '))      # 2 이상의 정수 입력
for i in range(2, num):           # i는 2부터 입력받은 수 num-1까지 반복
 if num % i==0:                   # i값으로 나누어떨어지는 수가 있다면
   break                          # 소수가 아니므로 반복문 탈출
if num == i + 1:                  # 1과 자신의 수 외 나누어 떨어진 수가 없다면
 print('소수')
else:                             # num이 어떤 수와 나누어떨어진다면
 print('소수아님')
```
