```python
def caloulator(a, operator, b):
    if operator == '+':
       return a + b
    elif operator == '-':
       return a - b
    elif operator == '*':
        return a * b
    elif operator == '/':
       if b == 0:
           return "오류: 0으로 나눌 수 없습니다."
       return a / b
    else:
       return "오류: 지원하지 않는 연산자입니다. (+, - *, /) 중 하나를 입력하세요"

# --사용 예시 --
print(caloulator(10, '+', 5))
```
