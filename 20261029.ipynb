```python
sum = 0                    # 합계를 저장할 변수 초깃값 설정
for i in range(1, 11, 1):  # i는 1부터 1씩 증가하여 10까지 변하고, 총 10회 반복
 sum += i                  # sum 값에 i의 값을 더함.
print('1부터 10까지의 합은: ', sum) # 결과 출력
```

```python
a = int(input('정수 입력: '))
sum = 0
for i in range(2, a+1,2):                   #i값이 짝수부터 시작하여 입력받은 값까지 짝수만 선택
 sum = sum + i                      # 짝수 값만 누적 계산
print('결과:', sum)
```
